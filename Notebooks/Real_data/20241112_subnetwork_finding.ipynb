```python
from biological_fuzzy_logic_networks.DREAM_analysis.utils import create_bfz
import networkx as nx
```

```python
model = create_bfz("/dccstor/ipc1/CAR/DREAM/DREAMdata/PKN_from_paper.sif", "dreambiofuzznet")
```

```python
root_nodes = ["ERK12", "MKK4", "MKK36", "SMAD23"]
def add_neighbors(l, n):
    for i in model.successors(n):
        if i in l:
            return l
        l.append(i)
        add_neighbors(l, i)
    return l

branches = {}
for i in model.successors("ERK12"):
    branches[i] = []
    for j in model.successors(i):
        branches[i]=add_neighbors(branches[i], j)
```

```python
root_nodes = ["ERK12", "MKK4", "MKK36", "SMAD23", "AKT", "GSK3B"]
def add_neighbors(l, n):
    for i in model.successors(n):
        if i in root_nodes:
            break
        l.append(i)
        l = add_neighbors(l, i)
    return l

branches = {}
for i in model.successors("ERK12"):
    branches[i] = []
    for j in model.successors(i):
        branches[i]=add_neighbors(branches[i], j)
```
